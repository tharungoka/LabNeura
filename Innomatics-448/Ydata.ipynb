```python
import ydata_profiling as yp
```

```python
import pandas as pd
```

```python
data = pd.read_csv("My_Uber_Drives_2016.csv")
```

```python
profiler = yp.ProfileReport(data)
```

```python
report = profiler.get_description()
```

```python
type(report)
```

```python
text_columns = list(map(lambda x:x[0],list(filter(lambda x: True if x[1]["type"]=="Text" else False,report.variables.items()))))
```

```python
categories = set(map(lambda x:x[1]["type"],report.variables.items()))
```

```python
categorical_columns = list(map(lambda x:x[0],list(filter(lambda x: True if x[1]["type"]=="Categorical" else False,report.variables.items()))))
```

```python
categorical_columns
```

```python
columns = {}
for category in categories:
    columns[category] = list(map(lambda x:x[0],list(filter(lambda x: True if x[1]["type"]==category else False,report.variables.items()))))
```
