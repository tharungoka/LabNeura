```python
y = 3 * x
```

```python
plt.plot(x,y,linestyle="-",color="green",linewidth=1)
plt.xticks(range(1,11))
plt.yticks(range(3,31))
plt.xlabel("X - Axis")
plt.ylabel("Y - Axis")
plt.title("Y = 3 * x ")
plt.savefig("before_show.png")
plt.show()
plt.savefig("after_show.png")
```
