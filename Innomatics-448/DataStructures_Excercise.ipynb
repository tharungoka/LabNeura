```python
# Given a list and a number.. we are finding whether the given number presnt in the list or not.
l1 = [1,2,3]
```

```python
# naive logic.
if l1[0] == 1:
    print("Found element 1 at 0")
elif l1[1] == 1:
    print("Found element 1 at 1")
elif l1[2] == 1:
    print("found element 1 at 2")
```

```python
# The above code suffers from bipolerate problem if we extend the list to large no of elements.
# To solve the above problem we have construct called for-loop.
```

```python
# inbuild function len
```

```python
"""
for <variable> in range(0,len(list)):
    <code block> 
    # Comparaision
    <code block>
"""
```

```python
# Bipolerate problem solved.

# Pointer / Variable based. Index based
for i in range(0,len(l1)):
    if l1[i] == 1:
        print(f"Element found at {i}")
```
