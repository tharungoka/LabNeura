```python
# To construct re-usable codes we use programming construct called functions.
# Components of the function
# Function Decleration
# Function Definition --> Explains us the name of the function, inputs, and outputs.
# Function Body
# Other sub components: inputs and outputs.
```

```python
"""
def <function_name>(input_list):
    //Function Body
    //outputting returning.
"""
```

```python
def get_minimum(l1):
    """
    Name of the function is : get_minimum
    Arguments:
        - l1: list
    Returns:
        - min_ : minimum value of list.
    """
    min_ = l1[0] # We should not use "min" as a variable name. (Not a good practice)
    for i in range(1,len(l1)): # for statement
        if l1[i] < min_: # if statement
            min_ = l1[i] # assignment statement
    return min_
```
