```python
def get_mean(arr):
    _sum = 0
    for ele in arr:
        # _sum = _sum + ele
        _sum += ele
    _mean = _sum/len(arr)
    return _mean
```

```python
l1 = [1,2,3,4,5,6]
```

```python
get_mean(l1)
```

```python
import math
def get_median(arr):
    sorted_arr = sorted(arr)
    n = len(arr)
    if n % 2 == 0:
        idx1 = int(n/2)
        idx2 = idx1 - 1
        _median = get_mean([arr[idx1],arr[idx2]])
        return _median
    else:
        idx = math.floor(n/2)
        _median = arr[idx]
        return _median
```

```python
## last frequency element will be updated
def get_mode(arr):
    dict_ = {}
    for i in arr:
        dict_[i] = 0
    for i in range(len(arr)):
        count_ = 0
        for j in range(len(arr)):
            if arr[i] == arr[j]:
                count_ += 1
        dict_[arr[i]] = count_
    new_dict = {value:key for key,value in dict_.items()}
    high_frequency = max(new_dict)
    mode_ = new_dict[high_frequency]
    return mode_
```

```python
class MatrixAdditionError(Exception):
    pass
    
class NewList(list):
    def __init__(self,l):
        super().__init__()
        self.list = list(l)
    def __add__(self,other):
        if len(self.list) == len(other.list):
            new_list = []
            for i in range(len(self.list)):
                new_list.append(self.list[i]+other.list[i])
            return new_list
        else:
            raise MatrixAdditionError("Addition is applicable only on same sized lists")
```

```python
def matmul_1(A,B,C):
    m,n = A.shape
    p,q = B.shape
    if n == p:
        #traversing rows in first matrix.. i is pointing to first row of A
        for i in range(m):
            #traversing columns in second matrix.. j is pointing to first column of B
            for j in range(q):
                #traversing column wise for rows of first matrix and row wise for columns of second matrix
                for k in range(n):
                    C[i][j] += A[i,k] * B[k,j]
```

```python
def matmul_2(A,B,C):
    m = A.shape[0]
    q = B.shape[1]
    for i in range(m):
        for j in range(q):
            # We removed traversal of elements and replaced with numpy vectorized multiplication.
            C[i][j] = np.sum(A[i]*B.T[j])
    # Time complexity is now m*q --> n**2
```
