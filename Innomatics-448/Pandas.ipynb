```python
def get_entry_between_9_10(args):
    start_time = args[0]
    end_time = args[1]
    idx = args[2]
    try:
        if (int(start_time.split(":")[0])==9 
                    and (int(end_time.split(":")[0])==9)):
            return idx
    except:
        print(start_time)
        return np.nan
```

```python
miles = df["MILES*"]
miles.shape
```

```python
min_ = miles.min()
max_ = miles.max()+1 # ()
```

```python
min_, max_
```

```python
bucket_size = (max_-min_)/5
```

```python
bucket_size
```

```python
buckets = {}
min_1 = min_
for i in range(5):
    buckets[f"bucket-{i+1}"] = {}
    buckets[f"bucket-{i+1}"]["start"] = min_1
    buckets[f"bucket-{i+1}"]["stop"] = min_1 + bucket_size
    buckets[f"bucket-{i+1}"]["count"] = 0
    min_1 += bucket_size
```

```python
buckets
```

```python
## Binning Algorithm
for mile in miles:
    for i in range(1,6):
        if (mile >= buckets[f"bucket-{i}"]["start"]) and (mile < buckets[f"bucket-{i}"]["stop"]):
            buckets[f"bucket-{i}"]["count"] += 1
```

```python
df["START_TIME*"] = df["START_DATE*"].apply(lambda x:x.split(" ")[1])
df["END_TIME*"] = df["END_DATE*"].apply(lambda x:x.split(" ")[1])
```

```python
def get_travelled_time(args):
    start_time = args[0]
    end_time = args[1]
    start_hrs = int(start_time.split(":")[0])
    start_mins = int(start_time.split(":")[1])
    end_hrs = int(end_time.split(":")[0])
    end_mins = int(end_time.split(":")[1])
    mins = (end_hrs - start_hrs) * 60 + (end_mins - start_mins)
    return mins
```

```python
df["TIME*"] = pd.Series(map(get_travelled_time, zip(df["START_TIME*"],df["END_TIME*"])))
```
