```python
l1 = [1,2,3,4,5,6]
```

```python
#extract even numbers
for ele in l1:
    if ele % 2 == 0:
        print(ele)
```

```python
def check_even_or_not(ele):
    if ele % 2 == 0:
        return True
    else:
        return False
```

```python
#extract even numbers
for ele in l1:
    even_status = check_even_or_not(ele)
    if even_status:
        print(ele)
```

```python
list(filter(check_even_or_not,l1))
```
