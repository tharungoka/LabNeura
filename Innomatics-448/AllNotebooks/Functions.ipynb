```python
def check_even_or_not_1(l1,l2):
    for i in l1:
        if i%2 == 0:
            l2.append(i)
```

```python
def check_even_or_not_2(l1):
    l2 = []
    for i in l1:
        if i%2 == 0:
            l2.append(i)
    return l2
```

```python
def check_even_or_not_3(x):
    if x % 2 == 0:
        return True
    else:
        return False
```

```python
#write a program to calculate the factorial of a number.
```

```python
n = 3
```

```python
factorial = 1
for i in range(n,0,-1):
    factorial *= i
```

```python
factorial
```

```python
def factorial(n):
    if n <= 1:
        return 1
    else:
        return  n*factorial(n-1)
```

```python
factorial(5)
```

```python
def inorder(node):
    if (node):
        inorder(node.left)
        print(node.data)
        inorder(node.right)
```

```python
def preorder(node):
    if (node):
        print(node.data)
        preorder(node.left)
        preorder(node.right)
```

```python
def postorder(node):
    if (node):
        postorder(node.left)
        postorder(node.right)
        print(node.data)
```
