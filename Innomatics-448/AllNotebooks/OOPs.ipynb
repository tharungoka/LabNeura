```python
#Constructor without parameters
class Human2:
    def __init__(self):
        self.ears = 2
        self.eyes = 2
```

```python
#Constructor with params
#Member variables/Instance Variables can be created inside the constructor.
#Member methods/Instance Methods can be created inside the class with self as a mandatory first argument.
#Class Variables can be created outside of constructor.
#Class Methods can be created using a decorator called @classmethod with an argument cls
class Human3:
    count = 0
    def __init__(self,ears,eyes,name):
        self.__ears = ears
        self._eyes = eyes
        self.name = name
        self.increment_count()
        self.__abc()

    @classmethod
    def increment_count(cls):
        cls.count += 1
        
    def eat(self):
        print(f"{self.name} is eating")

    def __abc(self):
        print("Private Function")

    def _cde(self):
        print("Protected Function")

    @classmethod
    def get_no_humans(cls):
        print(f"No of Humans: {cls.count}")
```

```python
human1 = Human1()
human2 = Human2()
```

```python
import math
class Pair:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __sub__(self,other):
        distance = math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2)
        return distance
    def __mul__(self,other):
        temp = (self.x * other.x) + (self.y * other.y)
        return temp
    def __copy__(self):
        return Pair(self.x,self.y)
```

```python
pair1 = Pair(1,2)
pair2 = Pair(2,3)
```

```python
pair1-pair2
```

```python
pair1 * pair2
```

```python
class MatrixAdditionError(Exception):
    pass
    
class NewList(list):
    def __init__(self,l):
        super().__init__()
        self.list = list(l)
    def __add__(self,other):
        if len(self.list) == len(other.list):
            new_list = []
            for i in range(len(self.list)):
                new_list.append(self.list[i]+other.list[i])
            return new_list
        else:
            raise MatrixAdditionError("Addition is applicable only on same sized lists")
```

```python
l3 = NewList([1,2])
l4 = NewList([1,2,3])
```

```python
l3 + l4
```
